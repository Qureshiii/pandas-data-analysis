Write a function that can return the track record of 2 teams against eachother.

```python
def track_record(ipl,x,y):
    match_filter = ((ipl['Team1'] == x) & (ipl['Team2'] == y)) | ((ipl['Team1'] == y) & (ipl['Team2'] == x))
    h2h = ipl[match_filter]
    
    total_matches = len(h2h)
    
    if total_matches == 0:
        return f"No matches found between {x} and {y}."
    
    team1_wins = (h2h['WinningTeam'] == x).sum()
    team2_wins = (h2h['WinningTeam'] == y).sum()
    no_result = total_matches - (team1_wins + team2_wins)
    
    
    stats = {
        'Total Matches': total_matches,
        f'{x} Wins': team1_wins,
        f'{y} Wins': team2_wins,
        'No Result / Ties': no_result,
        f'{x} Win %': round((team1_wins / total_matches) * 100, 2)
    }
    
    return stats
```

```python
track_record(ipl,'Royal Challengers Bangalore','Rajasthan Royals')
```
