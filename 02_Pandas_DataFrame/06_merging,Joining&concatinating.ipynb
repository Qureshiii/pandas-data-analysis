            IPL Problems

            Find top 3 stadiums with highest six/match ratio?

```python
temp_df = delivery.merge(matches,left_on='match_id',right_on='id')
six_df = temp_df[temp_df['batsman_runs'] == 6]
num_six_stadium = six_df.groupby('venue')['venue'].count()
num_match_stadium = matches['venue'].value_counts()

ratio = (num_six_stadium / num_match_stadium)

ratio.sort_values(ascending=False).head(3)
```

            Find orange cape holders of all season?

```python

temp_df.groupby(['season','batsman'])['batsman_runs'].sum().reset_index().sort_values('batsman_runs',ascending=False).drop_duplicates(subset=['season'],keep='first').sort_values('season')
```
