```python
# split , get // we are splitting first name ,surename , title.
df['last_name'] = df['Name'].str.split(',').str.get(0)
df[['title','first_name']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
```

            Find out how many Mr,Mrs,Ms,Miss.

```python
df['title'].value_counts()
```

            Want to sort out Miss,Ms and Mlle(french Miss) cuz this is a same category.

```python
# replace
df['title'] = df['title'].str.replace('Ms.','Miss.')
df['title'] = df['title'].str.replace('Mlle.','Miss.')
```

            Filtering 

```python
# startswith
df[df['first_name'].str.startswith('A')]
# endswith
df[df['first_name'].str.endswith('a')]
```

            filter applying regex(regular Expressions)

```python
# contains
# Search with both casses
df[df['first_name'].str.contains('John',case=False)]
# Find lastnames start and end with char vowel
df[df['first_name'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
# vowel ko chodh k kuch bhi ana chahiye(mtlb vowel k bina start and end waley names)
df[df['first_name'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```
