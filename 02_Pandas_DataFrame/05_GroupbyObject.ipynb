            find number of movies starting with A for each group?

```python
def movies_starts_with_A(group):
    return group['Series_Title'].str.startswith('A').sum()

genres.apply(movies_starts_with_A)
```

            Find ranking of each movie in the group according to IMDB score?

```python
def ranking(group):
    group['movie_rank'] = group['IMDB_Rating'].rank(ascending=False)

    return group

genres.apply(ranking)
```

```python
genres = movies.groupby('Genre')
```

            Find normalize IMDB rating group wise?

```python
def normalize(group):
    
    group['Normalize_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group

genres.apply(normalize)
```

            Find the top ten batsman in terms of runs?

```python
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

            Find the batsman with most number of sixes?

```python
six = ipl[ipl['batsman_runs'] == 6]
six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

            Find batsman with most numbers of 4's and 6's in last 5 overs?

```python

fours_six = ipl[(ipl['batsman_runs'] == 4) | (ipl['batsman_runs'] == 6) ]
last_overs = fours_six[fours_six['over'] > 15]
last_overs.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1)
```

            find virat kohli record against all teams?

```python
kohli = ipl[ipl['batsman'] == 'V Kohli']
kohli.groupby('bowling_team')['batsman_runs'].sum()
```

            create a function that can return the highest score of any batsman?

```python
def highest_score(batsman):
    temp = ipl[ipl['batsman'] == batsman]
    return temp.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]

highest_score('RG Sharma')
```
