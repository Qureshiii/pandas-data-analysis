```python
import numpy as np
import pandas as pd
```

Series From a list

```python
# string 
country = ['Pakistan','India','USA','Srilanka']
pd.Series(country)
```

```python
#Integer
runs = [13,24,56,70,100]
pd.Series(runs)
```

```python
#custom index
marks = [67,57,89,100]
subjects =['maths','English','Science','Urdu']
pd.Series(marks,index=subjects)
```

```python
#Setting a name
pd.Series(marks,index=subjects,name='Muttahir k Marks')
```
